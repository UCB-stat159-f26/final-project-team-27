## Inspect the Long-Format Granularity (Incidents vs. Subjects vs. Actions)
Goal: 
* inspect how many unique incidents and unique subjects exist compared to total rows
* inspect one example of a multi-row incident

```python
berkeley_df = df[df["City"] == "Berkeley"]
id_cols = ["Incident Number", "Subject_Incident Number", "Officer_Incident Number", "Use of Force Number"]

# inspect unique incidents
for col in id_cols:
    print(f"Unique {col}: {berkeley_df[col].nunique():,}")
    print(f"NAs in {col}:  {berkeley_df[col].isna().sum()}\n")
    
multi_inc = berkeley_df["Incident Number"].value_counts()
sample_id = multi_inc[multi_inc > 1].index[0]
berkeley_sample = berkeley_df[berkeley_df["Incident Number"] == sample_id]

print(f"\nExample multi-row incident #: {sample_id}, shape: {berkeley_sample.shape}")
display(berkeley_df[berkeley_df["Incident Number"] == sample_id])

example_multi_row_incident = berkeley_df[berkeley_df["Incident Number"] == sample_id]
for col in id_cols:
    print(f"Unique {col} in Incident #{sample_id}:  {example_multi_row_incident[col].nunique():,}")
    print(f"NAs in {col} in Incident #{sample_id}:  {example_multi_row_incident[col].isna().sum()}\n")
```

***Some Things to Note:***
1. Missing Subject_Incident Numbers exists
2. Multi-subject incidents are cross-joined:
   * In Incident #4442, there are 17 unique Use of Force Numbers and 3 unique Subject_Incident Numbers, which equals $17 \times 3 = 51$ rows
   * When an incident involves multiple people, BPD’s export pairs every force action with every subject in that incident rather than linking a specific officer's action to one specific person
   * Cannot assume a specific force action in a multi-subject incident was applied to that row's subject
   * This ambiguity should not be too much of a problem because most incidents only have 1 person (1,788 incidents & 1,727 subjects, meaning ~95% of incidents only involve a single person ie. know with 100% certainty that person received every listed force action
   * ***TODO:*** check how many single vs multple subjects incidents 
3. Multiple actions per officer are logged separately
    * Officer 10421 appears multiple times in Incident #4442 under distinct Use of Force Numbers
    * Each distinct technique deployed during an incident generates its own Use of Force Number
4. Subject Role complements Subject Arrested:
    * The Subject Role column distinguishes between "Arrested" and "Involved Party"

## Inspect Missing Values and Previous TODOs Mentioned Above

```python
# Missing values per column (in Berkeley data)
print("--- Missing Values ---")
print(berkeley_df.isna().sum()[berkeley_df.isna().sum() > 0])

# Check date range and NonBerkeley Address
print("\n--- Date Range ---")
dates = pd.to_datetime(berkeley_df["OccurredDate"], errors="coerce")
print(f"Min date: {dates.min()} | Max date: {dates.max()}")

print("\n--- NonBerkeley Address Values ---")
print(berkeley_df["NonBerkeley Address"].value_counts(dropna=False))

# Check how many incidents have 1 subject vs. multiple subjects
subj_per_inc = berkeley_df.groupby("Incident Number")["Subject_Incident Number"].nunique()
print("\n--- Subjects per Incident ---")
print(subj_per_inc.value_counts().sort_index())

# Cross-tab Subject Role vs. Subject Arrested (deduplicated per subject)
unique_subjects = berkeley_df.drop_duplicates(subset=["Incident Number", "Subject_Incident Number"])
print("\n--- Subject Role vs. Subject Arrested (Unique Subjects) ---")
print(pd.crosstab(unique_subjects["Subject Role"], unique_subjects["Subject Arrested"], dropna=False))

# Check unique values for outcome & demographic columns
print("\n--- Unique Values Check ---")
outc_demo_cols = ["Use of Force Level", "Use of Force Type", "Use of Force Reason", "Subject Arrested", "Race", "Sex"]

for col in outc_demo_cols:
    print("\n", berkeley_df[col].value_counts(dropna=False))
```
